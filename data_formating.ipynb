```python
def rev_extra(txt):
    new_txt = ''
    for c in txt.split():
        if c == 'Customer' or c == ':' or c == 'Agent': continue
        else: new_txt = new_txt + c + " "

    return new_txt[1:]

print(rev_extra('Customer :I remember going to see the fireworks with my best friend. It was the first time we ever spent time alone together. Although there was a lot of people, we felt like the only people in the world. Agent :'))
```

```python
data['empathetic_dialogues'] = data['empathetic_dialogues'].apply(rev_extra)
```
